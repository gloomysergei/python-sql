####  `Менеджер для соединения с PostgreSQL (с commit/rollback)`

```python
import psycopg

class PostgresConnection:
    def __init__(self, conn_str):
        self.conn_str = conn_str
        self.conn = None

    def __enter__(self):
        self.conn = psycopg.connect(self.conn_str)
        return self.conn

    def __exit__(self, exc_type, exc_val, exc_tb):
        try:
            if exc_type is None:
                # если ошибок не было — фиксируем транзакцию
                self.conn.commit()
            else:
                # если была ошибка — откатываем транзакцию
                self.conn.rollback()
        finally:
            # в любом случае закрываем соединение
            if self.conn:
                self.conn.close()
```

##### `Использование:`

```python
with PostgresConnection("postgresql://sergey@/hexlet?host=/var/run/postgresql") as conn:
    with conn.cursor() as cur:
        cur.execute("SELECT * FROM users;")
# здесь: commit/rollback уже выполнен, соединение закрыто
```
